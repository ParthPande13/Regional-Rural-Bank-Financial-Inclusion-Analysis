```python
# Remove duplicate records
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
print("Duplicate rows after cleaning:", df.duplicated().sum())
```

```python
# Reset index after dropping duplicates
df = df.reset_index(drop=True)
```

```python
# Check Record_ID uniqueness post-cleaning
print("Unique Record_IDs:", df["Record_ID"].nunique(), " Total rows:", len(df))
```

```python
# Checking missing values again after dedup
df.isnull().sum()
```

```python
# Remove rows with missing values
# droping because its only 20rows 1% of 2000
df = df.dropna()

print("Shape after removing missing values:", df.shape)
print("Missing values after cleaning:", df.isnull().sum().sum())
```

```python
# --- Feature Engineering / Transformation ---

# Account-level ratios
df["Active_Ratio"] = df["Active_Accounts"] / df["Total_Accounts"]
df["Women_Ratio"] = df["Women_Accounts"] / df["Total_Accounts"]
df["JanDhan_Ratio"] = df["Jan_Dhan_Accounts"] / df["Total_Accounts"]

# Digital adoption ratios
df["Mobile_Banking_Ratio"] = df["Mobile_Banking_Users"] / df["Total_Accounts"]
df["Internet_Banking_Ratio"] = df["Internet_Banking_Users"] / df["Total_Accounts"]
df["UPI_Share_of_Digital"] = df["UPI_Transactions"] / df["Digital_Transactions"]

# Loan composition ratios
df["Agri_Loan_Share"] = df["Agricultural_Loans_Lakh"] / df["Total_Loans_Lakh"]
df["MSME_Loan_Share"] = df["MSME_Loans_Lakh"] / df["Total_Loans_Lakh"]
df["Personal_Loan_Share"] = df["Personal_Loans_Lakh"] / df["Total_Loans_Lakh"]

# Deposit composition ratios
df["Savings_Deposit_Share"] = df["Savings_Deposits_Lakh"] / df["Total_Deposits_Lakh"]
df["Term_Deposit_Share"] = df["Term_Deposits_Lakh"] / df["Total_Deposits_Lakh"]

# Outreach / infrastructure metrics
df["Accounts_per_Branch"] = df["Total_Accounts"] / df["Branches"]
df["Population_per_Branch"] = df["Population"] / df["Branches"]
df["Population_per_Outlet"] = df["Population"] / df["Banking_Outlets"]
df["Deposits_per_Account"] = df["Total_Deposits_Lakh"] / df["Total_Accounts"]
df["Loans_per_Account"] = df["Total_Loans_Lakh"] / df["Total_Accounts"]

# Time feature: combined Year-Quarter for trend analysis
df["Year_Quarter"] = df["Year"].astype(str) + "-" + df["Quarter"].astype(str)


print("Shape after transformation:", df.shape)
df.head()
```

```python
import os
os.makedirs("data/processed", exist_ok=True)

df.to_csv("data/processed/rrb_cleaned.csv", index=False)
```

```python

# --- Exploratory Analysis: State-wise ---

state_summary = df.groupby("State").agg(
    Total_Accounts=("Total_Accounts", "sum"),
    Avg_Active_Ratio=("Active_Ratio", "mean"),
    Avg_Women_Ratio=("Women_Ratio", "mean"),
    Avg_Digital_Adoption=("Mobile_Banking_Ratio", "mean"),
    Avg_Accounts_per_Branch=("Accounts_per_Branch", "mean"),
    Total_Deposits=("Total_Deposits_Lakh", "sum"),
    Total_Loans=("Total_Loans_Lakh", "sum"),
).sort_values("Total_Accounts", ascending=False)

state_summary

# --- Region-wise ---

region_summary = df.groupby("Region").agg(
    Total_Accounts=("Total_Accounts", "sum"),
    Avg_Active_Ratio=("Active_Ratio", "mean"),
    Avg_Women_Ratio=("Women_Ratio", "mean"),
    Avg_JanDhan_Ratio=("JanDhan_Ratio", "mean"),
    Avg_Digital_Adoption=("Mobile_Banking_Ratio", "mean"),
).sort_values("Total_Accounts", ascending=False)

region_summary

# --- Rural vs Urban vs Semi-Urban ---

rural_urban_summary = df.groupby("Rural_Urban").agg(
    Avg_Active_Ratio=("Active_Ratio", "mean"),
    Avg_Women_Ratio=("Women_Ratio", "mean"),
    Avg_JanDhan_Ratio=("JanDhan_Ratio", "mean"),
    Avg_Digital_Adoption=("Mobile_Banking_Ratio", "mean"),
    Avg_Population_per_Branch=("Population_per_Branch", "mean"),
    Avg_Accounts_per_Branch=("Accounts_per_Branch", "mean"),
)

rural_urban_summary

# --- Year-wise trend (numeric summary, not chart yet) ---

year_summary = df.groupby("Year").agg(
    Total_Accounts=("Total_Accounts", "sum"),
    Avg_Digital_Adoption=("Mobile_Banking_Ratio", "mean"),
    Total_Digital_Transactions=("Digital_Transactions", "sum"),
    Total_UPI_Transactions=("UPI_Transactions", "sum"),
)

year_summary

# ratio columns 

ratio_cols = [
    "Active_Ratio", "Women_Ratio", "JanDhan_Ratio",
    "Mobile_Banking_Ratio", "Internet_Banking_Ratio", "UPI_Share_of_Digital",
    "Agri_Loan_Share", "MSME_Loan_Share", "Personal_Loan_Share",
    "Savings_Deposit_Share", "Term_Deposit_Share",
    "Accounts_per_Branch", "Population_per_Branch", "Population_per_Outlet",
    "Deposits_per_Account", "Loans_per_Account"
]

ratio_corr = df[ratio_cols].corr()

# --- Correlation matrix (numbers only, heatmap comes later) ---

corr_matrix = df.select_dtypes(include=np.number).corr()
corr_matrix["Active_Ratio"].sort_values(ascending=False)
```

```python
# Visualization: Digital Adoption vs Active Ratio 

plt.figure(figsize=(8, 6))
sns.scatterplot(x="Mobile_Banking_Ratio", y="Active_Ratio", data=df, alpha=0.5)
sns.regplot(x="Mobile_Banking_Ratio", y="Active_Ratio", data=df, scatter=False, color="red")
plt.title("Digital Adoption vs Account Activity")
plt.xlabel("Mobile Banking Ratio")
plt.ylabel("Active Account Ratio")
plt.show()

```

```python
# --- State-wise Total Accounts 

plt.figure(figsize=(10, 6))
sns.barplot(x=state_summary["Total_Accounts"], y=state_summary.index)
plt.title("Total Accounts by State")
plt.xlabel("Total Accounts")
plt.show()
```

```python
# Rural vs Urban vs Semi-Urban: Active Ratio & Digital Adoption 

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.barplot(x=rural_urban_summary.index, y=rural_urban_summary["Avg_Active_Ratio"], ax=axes[0])
axes[0].set_title("Avg Active Ratio by Area Type")

sns.barplot(x=rural_urban_summary.index, y=rural_urban_summary["Avg_Digital_Adoption"], ax=axes[1])
axes[1].set_title("Avg Digital Adoption by Area Type")
plt.tight_layout()
plt.show()
```

```python
#  Correlation heatmap 

plt.figure(figsize=(12, 10))
sns.heatmap(ratio_corr, cmap="coolwarm", center=0, annot=True, fmt=".2f", 
            annot_kws={"size": 8}, linewidths=0.5)
plt.title("Correlation Heatmap - Financial Inclusion Ratios")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

```python
# Active Ratio trend over Year

year_active = df.groupby("Year")["Active_Ratio"].mean()

plt.figure(figsize=(8, 5))
sns.lineplot(x=year_active.index, y=year_active.values, marker="o")
plt.title("Average Active Account Ratio by Year")
plt.xlabel("Year")
plt.ylabel("Avg Active Ratio")
plt.show()
```
